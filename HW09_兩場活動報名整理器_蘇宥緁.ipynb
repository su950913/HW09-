```markdown
# HW09 兩場活動報名整理器

在開始執行程式之前，我們先來釐清重要的集合概念，並試著預測看看測試資料的執行結果！

### 1. 集合（Set）概念與「人數 vs. 人次」說明

* **交集 (Intersection `&`)**：同時報名「活動 A」且報名「活動 B」的名單。代表兩場都想參加的人。
* **聯集 (Union `|`)**：報名「活動 A」或報名「活動 B」的名單（去重後）。代表這次活動總共服務了多少個「不同」的人。
* **差集 (Difference `-`)**：
  * **A - B (A 差集 B)**：只有報名「活動 A」，但沒有報名「活動 B」的名單。
  * **B - A (B 差集 A)**：只有報名「活動 B」，但沒有報名「活動 A」的名單。
* **人數 (Unique Users) vs. 人次 (Total Registrations)**：
  * **人次**：不排除重複。如果同一個人報名了活動 A 也報名了活動 B，或是單一活動重複報名，每次報名都算一次（例如：活動 A 報名 5 次 + 活動 B 報名 4 次 = 9 人次）。
  * **人數**：排除重複。透過 `set` 進行去重，計算實際上有多少個「獨立的個體」（不重複的姓名）參與。

---

### 2. 測試資料與預測引導

假設我們有以下兩組原始報名資料：
* **活動 A 原始報名**：`" alice , Bob,  alice, CHARLIE "`
* **活動 B 原始報名**：`"BOB, Charlie, David,  bob "`

**步驟一：清理與分割後的 List（保留重複）**
* 請試著將它們轉為全小寫、去除多餘半形空白，並分割成 List：
  * 活動 A List 預測：`['alice', 'bob', 'alice', 'charlie']`
  * 活動 B List 預測：`['bob', 'charlie', 'david', 'bob']`

**步驟二：轉成 Set（自動去重，用於收費計算）**
* 活動 A Set 預測：`{'alice', 'bob', 'charlie'}`
* 活動 B Set 預測：`{'bob', 'charlie', 'david'}`
```

````python
# =====================================================================
# 第一組測試：基本測試資料
# =====================================================================

# 依規格：桌遊體驗（活動 A）每人 120 元，電影欣賞（活動 B）每人 80 元
fee_a = 120
fee_b = 80

# 原始報名資料 (含有大小寫不一、前後多餘空白、以及重複報名)
raw_data_a = " ben, Dora, ERIC, ben "
raw_data_b = "BOB, Charlie, David,  bob "

# ---------------------------------------------------------------------
# 步驟 1：資料清理與分割 (將大小寫一致化、去除多餘空白，轉為 List 保留重複)
# ---------------------------------------------------------------------
# 先將整串字串轉為小寫
lower_a = raw_data_a.lower()
lower_b = raw_data_b.lower()

# 去除所有的半形空白
clean_a = lower_a.replace(" ", "")
clean_b = lower_b.replace(" ", "")

# 使用逗號進行分割，得到包含重複名單的 List
list_a = clean_a.split(",")
list_b = clean_b.split(",")

# ---------------------------------------------------------------------
# 步驟 2：利用 Set 進行去重 (用於計費與集合運算)
# ---------------------------------------------------------------------
set_a = set(list_a)
set_b = set(list_b)

# ---------------------------------------------------------------------
# 步驟 3：集合運算 (交集、聯集、雙向差集)
# ---------------------------------------------------------------------
# 交集：兩場都報名的人
both_events = set_a & set_b

# 聯集：有報名任一場的人（總不重複人數）
any_event = set_a | set_b

# 差集 A - B：只有報名 A 沒報名 B 的人
only_a = set_a - set_b

# 差集 B - A：只有報名 B 沒報名 A 的人
only_b = set_b - set_a

# ---------------------------------------------------------------------
# 步驟 4：人數與人次計算、收費統計
# ---------------------------------------------------------------------
# 各自去重後的人數
unique_a = len(set_a)
unique_b = len(set_b)

# 依規格定義：
# 「報名人次」指兩場活動各自去重後的人數相加
total_registrations = unique_a + unique_b

# 「總不重複人數」指至少參加一場活動的人數 (即聯集數量)
total_unique_people = len(any_event)

# 收費計算 (以去重後的人數乘以單價計算)
income_a = unique_a * fee_a
income_b = unique_b * fee_b
total_income = income_a + income_b

# ---------------------------------------------------------------------
# 步驟 5：格式化輸出結果 (依規格：去重名單需按字母順序排序)
# ---------------------------------------------------------------------
print(f"=== 第一組測試結果 ===")
print(f"桌遊體驗 清理後的原始名單: {list_a}")
print(f"電影欣賞 清理後的原始名單: {list_b}")
print(f"桌遊體驗 排序去重名單 (收費依據): {sorted(list(set_a))}")
print(f"電影欣賞 排序去重名單 (收費依據): {sorted(list(set_b))}")
print("-" * 40)
print(f"兩場都報名的人 (交集): {sorted(list(both_events))}")
print(f"總不重複人數名單 (聯集): {sorted(list(any_event))}")
print(f"僅報名桌遊體驗的人 (A 差集 B): {sorted(list(only_a))}")
print(f"僅報名電影欣賞的人 (B 差集 A): {sorted(list(only_b))}")
print("-" * 40)
print(f"報名人次: {total_registrations} 人次 (桌遊: {unique_a}, 電影: {unique_b})")
print(f"總不重複人數: {total_unique_people} 人")
print(f"桌遊體驗預計收入: {income_a} 元")
print(f"電影欣賞預計收入: {income_b} 元")
print(f"總預計實收金額: {total_income} 元")
````

````python
# =====================================================================
````
````python
# 最終作業產出：使用環境變數與規格要求的正式整理器
# =====================================================================

# 固定的兩個活動名稱，以 tuple 保存。
activity_names = ("桌遊體驗", "電影欣賞")

# 收到的原始文字名單，請保留原值。
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " AMY, ben, CORA, amy "

# 用活動名稱查詢每人費用。
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# ---------------------------------------------------------------------
# 步驟 1：字串清理（大小寫一致化、去除空白、並分割成 List）
# ---------------------------------------------------------------------
name_board = activity_names[0]
name_movie = activity_names[1]

fee_board = float(unit_fees[name_board])
fee_movie = float(unit_fees[name_movie])

# 清理字串 (全小寫、取代半形空白為空字元)
clean_board_str = board_raw.lower().replace(" ", "")
clean_movie_str = movie_raw.lower().replace(" ", "")

# 分割成原始報名 List (保留重複)
list_board = clean_board_str.split(",")
list_movie = clean_movie_str.split(",")

# ---------------------------------------------------------------------
# 步驟 2：利用 Set 進行去重
# ---------------------------------------------------------------------
set_board = set(list_board)
set_movie = set(list_movie)

# ---------------------------------------------------------------------
# 步驟 3：集合運算與排序
# ---------------------------------------------------------------------
# 交集、聯集、雙向差集
both_intersection = set_board & set_movie
all_union = set_board | set_movie
only_board_diff = set_board - set_movie
only_movie_diff = set_movie - set_board

# 排序輸出名單 (符合按小寫英文字母順序排序規則)
sorted_board = sorted(list(set_board))
sorted_movie = sorted(list(set_movie))
sorted_intersection = sorted(list(both_intersection))
sorted_union = sorted(list(all_union))
sorted_only_board = sorted(list(only_board_diff))
sorted_only_movie = sorted(list(only_movie_diff))

# ---------------------------------------------------------------------
# 步驟 4：計算人數、人次、通知名單與費用
# ---------------------------------------------------------------------
unique_count_board = len(set_board)
unique_count_movie = len(set_movie)

# 報名人次：兩場活動各自去重後人數相加
total_registrations = unique_count_board + unique_count_movie

# 總不重複人數：至少參加一場活動的人數 (聯集大小)
total_unique_people = len(all_union)

# 通知名單前兩位：全部參加者排序後的前兩位
notice_list_top2 = sorted_union[:2]

# 兩場分別計費與總金額 (以 float 格式運算)
income_board = unique_count_board * fee_board
income_movie = unique_count_movie * fee_movie
total_income = income_board + income_movie

# ---------------------------------------------------------------------
# 步驟 5：結果輸出呈現 (符合作業期望格式)
# ---------------------------------------------------------------------
print("HW09 活動報名摘要")
print(f"活動：{name_board}，每人 {fee_board:.2f} 元")
print(f"活動：{name_movie}，每人 {fee_movie:.2f} 元")
print(f"{name_board[:2]}整理串列：{list_board}")
print(f"{name_movie[:2]}整理串列：{list_movie}")
print(f"{name_board[:2]}去重名單：{sorted_board}")
print(f"{name_movie[:2]}去重名單：{sorted_movie}")
print(f"{name_board[:2]}人數：{unique_count_board} 人")
print(f"{name_movie[:2]}人數：{unique_count_movie} 人")
print(f"共同參加者：{sorted_intersection}")
print(f"共同人數：{len(both_intersection)} 人")
print(f"只參加{name_board[:2]}者：{sorted_only_board}")
print(f"只參加{name_movie[:2]}者：{sorted_only_movie}")
print(f"全部參加者：{sorted_union}")
print(f"總不重複人數：{total_unique_people} 人")
print(f"報名人次：{total_registrations} 人次")
print(f"通知名單前兩位：{notice_list_top2}")
print(f"{name_board[:2]}應收：{income_board:.2f} 元")
print(f"{name_movie[:2]}應收：{income_movie:.2f} 元")
print(f"總應收：{total_income:.2f} 元")
````

1. 為什麼桌遊費用要以3 人計算？
原始資料狀況：在測試一中，桌遊體驗整理後的原始名單為['amy', 'ben', 'amy', 'cora']，這裡面總共有4 筆報名紀錄。
重複報名規則：名單中的'amy'重複出現了兩次。根據活動主辦方的規則：「同一個人在同一場活動中重複出現，只算一位參加者，也只收一次該場費用。」
實際收費人數：剔除重複的'amy'之後，實際報名的不重複人員只有'amy'、'ben'、'cora'這3 個人。
計費公式：因此，費用必須以實際不重複的3 人來計算，即：3 人× 120 元= 360 元。

2. 直接把字符串拆成List 保留與失去了什麼？
保留了：重複性。每一個報名紀錄都會完整保留下來（例如可以知道'amy'報名了兩次）。
保留了：順序性。元素會完全依照使用者報名的順序排列。
失去了：唯一性。串列無法自動幫我們過濾重複的姓名，因此不能直接拿來當作計算「實際到場人數」與「應收費用」的依據。

3. 再將List 轉換成Set 保留與失去了什麼？
保留了：唯一性(去重)。集合會自動刪除重複的元素，只留下唯一的不重複姓名，非常適合直接拿來計算實際應收費用的「總人數」。
失去了：重複紀錄。我們無法再得知哪些姓名曾經重複報名過、重複了幾次（失去了「人次」的歷程軌跡）。
失去了：順序性。集合在Python 中是無序的，我們無法得知這些人在原始資料中誰先報名、誰後報名。
